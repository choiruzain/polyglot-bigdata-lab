# Hello Spark
Run `sh scripts/platform.sh load-sample` once first. It loads the shop dataset into Hive, and this notebook reads it through the shared metastore.

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('hello').enableHiveSupport().getOrCreate()
spark.sql('show tables in shop').show()
spark.sql('select category, count(*) as products from shop.products group by category order by category').show()
```

```python
spark.sql("""
    SELECT *
    FROM products
    WHERE category = 'Books'
""").show()
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("t3-hive").enableHiveSupport().getOrCreate()
print("DATABASES", [r[0] for r in spark.sql("show databases").collect()])
print("TABLES", [r.tableName for r in spark.sql("show tables in shop").collect()])
print("ORDERS", spark.sql("select count(*) from shop.orders").first()[0])
spark.createDataFrame([(3, "carol"), (4, "dave")], ["id", "name"]) \
     .write.mode("overwrite").saveAsTable("spark_demo")
print("SPARK_DEMO", spark.sql("select * from spark_demo order by id").collect())
spark.stop()
```
